```python
#lista

num = [1,2,3,4,5]

print(num[2])
```

```python
#for

for i in num:
    if i%2 == 0:
        print(f"o número {i} é par\n")
```

```python
# +=

soma_par = 0

for i in num:
    if i % 2 == 0:
        soma_par += i

print(f"a soma nos números pares da lista é: {soma_par}")
```

```python
# função
def soma_par (lista: list) -> tuple[int, list]:
    soma_par = 0
    lista_par = []    
    for i in lista:
        if i % 2 == 0:
            soma_par += i
            lista_par.append(i)
    return soma_par, lista_par

somapar, listapar = soma_par(num)

print(f"a soma dos números pares na lista {listapar} é {somapar}")

```

```python
# dicionarios
'''
pessoas = [
    {"":"", "":""}
    {"":"", "":""}
    {"":"", "":""}
]
'''

pessoas = [
    {"name":"ana", "age":24},
    {"name":"bob", "age":22},
    {"name":"cal", "age":23}
]

print(pessoas[1])
print(pessoas[1]["age"])
```

```python
def media_idade (lista:list)->int:
    soma = 0
    for i in pessoas:
        soma += i["age"]
    return soma/len(pessoas)

print(f"a média de idade das pessoas na lista é: {media_idade(pessoas)}")
```

```python
# ler da web

import requests
import csv
import io 

url = 'https://raw.githubusercontent.com/ccalmendra/ciencia-dados/refs/heads/main/dados/m1-pessoas.csv'

response = requests.get(url)

reader_csv = csv.reader(response.text.splitlines()) # splitlines() trasnforma o textão concatenado em várias listas

header = next(reader_csv) # a primeira linha é o nome das colunas, o next pula ela e salva ela numa var chamada header

idades=[]
idx_idade = header.index('idade') # pego o index da coluna que tem nome idade

for linha in reader_csv:
    if linha:
        idades.append(int(linha[idx_idade])) # coloco a idade numa lista usando o index que achamos acima, tipe pois pode não ser int !!!!

media = sum(idades)/len(idades)

print(media)
```

```python
#csv

import json

estudantes = [
    {"nome": "Ana", "notas": [8, 7, 9]},
    {"nome": "Carlos", "notas": [6, 5, 8]},
    {"nome": "Mariana", "notas": [9, 10, 10]}
]

# criar o json
with open('m1-estudantes.json', 'w') as jsonfile:
    json.dump(estudantes, jsonfile, indent=4, ensure_ascii=False) #pega o objeto estudantes e escreve nessa var jsonfile, indent 4 deixa organizado e o ensure_ascii = False deixa acentos

#ler e atualizar

with open('m1-estudantes.json', 'r') as jsonfile:
    dados = json.load(jsonfile) # lê o json do arquivo e tranforma ele em dados
    for estudante in dados:
        estudante['media'] = sum(estudante['notas']) / len(estudante['notas'])#criamos uma nova chave

#pra salvar tudo
with open('m1-estudantes_atualizado.json', 'w') as jsonfile:
    json.dump(dados, jsonfile, indent=4)
```

```python
import requests
import nltk #natural language toolkit, processa linguagem natural que analisa o texto 
from nltk.tokenize import word_tokenize # divide texto em palavras
from collections import Counter # conta freq de elementos

nltk.download('punkt') #modelos pra funcionar, se não o word_tokenize não funciona 
nltk.download('punkt_tab')

# URL do arquivo de texto não estruturado
url = "https://raw.githubusercontent.com/ccalmendra/ciencia-dados/refs/heads/main/dados/m2-spam.txt"

# Fazer a requisição HTTP e obter o texto
try:
    response = requests.get(url)
    response.raise_for_status() # Levanta um erro para códigos de status ruins (4xx ou 5xx)
    texto = response.text

    tokens = word_tokenize(texto.lower()) #tokeniza nosso texto em minusculo
    contagem_palavras = Counter(tokens) #conta quantos tokes temos de cada palavra
    print(contagem_palavras)
except requests.exceptions.RequestException as e:
    print(f"Erro ao carregar o texto da URL: {e}")
```

```python
# Criando um array com as notas dos alunos
notas = np.array([
        [7.5, 8.0, 6.0],
        [5.5, 6.0, 8.0],
        [9.0, 7.5, 8.5],
        [4.5, 6.5, 7.0],
        [8.0, 8.5, 9.0]])

# Média por aluno (média ao longo das colunas)
media_alunos = np.mean(notas, axis=1) # axis = 0 iria fazer a media com as colunas somando todos os elementos de uma coluna
media_geral = np.mean(notas) #poderiamos ter um keepdims = True aqui para não sumir as dimensões

print("Notas dos alunos:\n", notas)
print("Média de cada aluno:", media_alunos)
print("Média geral:", media_geral)
```

```python
# matriz bi com 5 colunas e 10 linhas 
matriz = np.random.randint(1, 51, (10, 5))

#calculando estatisticas dentro das linhas 
means = np.mean(matriz, axis=1, keepdims=True)
medians = np.median(matriz, axis=1, keepdims=True)
stds = np.std(matriz, axis=1, keepdims=True)

matriz_nova = np.concatenate((matriz, means, medians, stds), axis=1)



print("Matriz original:")
print(matriz)
print("\nMatriz com estatísticas calculadas:")
print(matriz_nova)


# Dica 1: use o recurso de realizar uma chamada de função ou operação em uma linha inteira para calcular as estatísticas
# Dica 2: use a função 'concatenate' para inserir várias novas colunas
# Dica 3: usar reshape para tornar o array de médias/medianas/desvios em colunas
```

```python
# estatisticas

import pandas as pd

df = pd.read_csv("https://github.com/ccalmendra/ciencia-dados/releases/download/lcd-v1/m3-atletas-eventos-olimpiadas.csv")

print(df.head())

df.describe()
```

```python
qtd_nomes_unicos = df["Name"].nunique()
print(qtd_nomes_unicos)

#mas posso ter nomes iguais de diferentes atletas -> Mohamed, Wang, Li, João, John

qtd_ids_unicos = df["ID"].nunique()
print(qtd_ids_unicos)

```

```python
# agregação simples goupby()[]

participacao_atletas = df.groupby("Name")["Event"].count() # name vira o index e event vira a coluna da series depois de aplicar a função
# o () def os grupos que viraram idx depois da agregação,  o [] def qual coluna sera analisa dentro de cada grupo, no nosso caso
# a coluna eventos é analisada dentro de cada nome 
print(participacao_atletas)

```

```python
id_atleta_mais_eventos = participacao_atletas.idxmax()
max_eventos = participacao_atletas.max()
print(id_atleta_mais_eventos)
print("participou de ",max_eventos,"eventos")
```

```python
atleta_mais_jogos = df.groupby("Name")["Games"].nunique() # o count contaria duas vezes o mesmo jogo pro caso do robert como se fossem jogos diferentes, fazemos um nunique pra pegar so a quantidade unica de nome dos jogos

id_maisjogos = atleta_mais_jogos.idxmax()

print(id_maisjogos)

# df[linha][colunas]  -> podemos filtrar subseries/dataframes com isso 

jogos = df[ df["Name"] == id_maisjogos ] ["Games"].unique()
print(jogos)
```

```python
# esportes com os 25% menos participações

quartil_25 = df['Sport'].value_counts().quantile(0.25) #ao contrario do count que contaria todo valor nao nulo o .value_counts() conta a frequencia de cada valor unico 
#o quantile pegaria o VALOR do quartil 25% entre as frequencias de esportes tipo os esportes que foram jogados 1, 2, 3 vezes

esportes_abaixoq25 = df['Sport'].value_counts() [df['Sport'].value_counts() <= quartil_25]
#o df_atletas_eventos['Sport'].value_counts() cria uma series com nome do sport no idx e a contagem dele na coluna
# o [] depois disso filtra essa serie só abaixo do q25

print(esportes_abaixoq25)

df_novo = df['Sport'].value_counts()

df_novo = df_novo[df_novo <= quartil_25] # df[df com condição]

print (df_novo)
```

```python
paises = df.groupby(["NOC", "Medal"]).size()
print("antes do unstack temos\n", paises)


paises = df.groupby(["NOC", "Medal"]).size().unstack(fill_value=0) # precisamos colocar o unstack(fill_value=0)

print("\napos o unstack fica:\n ", paises)

paises_sem_ouro = paises[paises["Gold"]==0] # só trabalhamos com o gold pois ele vira coluna com o unstack

total_medalhas = paises_sem_ouro.sum(axis=1) #o sum ocorre dentro das medalhas do país, retorna uma series com o nome do pais e o total de medalhas
#o df paises sem ouro tem como indice team e medal, o axis = 1 trabalha dentre da linha, horizontalmente
#sem axis =1 ele somaria todos os paises 
indice = total_medalhas.idxmax() # pegamos o indice daquele que é o maior no total de medalhas
valor_medalhas = total_medalhas.max() #o numero delas

print("O país com mais medalhas sem ter ouro é: ", indice, "com cerca de ", valor_medalhas, "medalhas")

top5 = total_medalhas.nlargest(5)
print(top5)
```

```python
# 1. Criamos um DataFrame com a contagem de medalhas por Time e Tipo
contagem = df.groupby(["Team", "Medal"]).size().unstack(fill_value=0)

# 2. Filtramos: Queremos apenas países onde a coluna 'Gold' seja igual a 0
sem_ouro = contagem[contagem["Gold"] == 0]

# 3. Agora somamos todas as medalhas (Silver + Bronze) para esses países
# e pegamos o índice (nome do país) do maior valor
total_medalhas_sem_ouro = sem_ouro.sum(axis=1)
indice_pais = total_medalhas_sem_ouro.idxmax()
valor_medalhas = total_medalhas_sem_ouro.max()

print(f"O time com mais medalhas sem ter ouro é: {indice_pais} com {valor_medalhas} medalhas.")

top5 = total_medalhas_sem_ouro.nlargest(5)
print(top5)
```

```python
# Separação por sexo
df_atletas_eventos_m = df[df['Sex'] == 'M']
df_atletas_eventos_f = df[df['Sex'] == 'F']

# Calcular Q1, Q3 e IQR - Feminino
q1 = df_atletas_eventos_f['Age'].quantile(0.25)
mean = df_atletas_eventos_f['Age'].median()
q3 = df_atletas_eventos_f['Age'].quantile(0.75)
iqr = q3 - q1 # Intervalo Interquartil

# Calcular os limites do IQR - Feminino
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

# Identificar os outliers - Feminino
outliers_superior = df_atletas_eventos_f[(df_atletas_eventos_f['Age'] > limite_superior)]
outliers_inferior = df_atletas_eventos_f[(df_atletas_eventos_f['Age'] < limite_inferior)]

# Exibir os resultados - Feminino
print(f"FEMININO\nQ1: {q1}, \nMedian: {mean}, \nQ3: {q3}, \nIQR: {iqr}")
print(f"Limite Inferior: {limite_inferior}, \nLimite Superior: {limite_superior}")
#print("Outliers:")
print(outliers_inferior[['ID','Sex','Age']])
print(outliers_superior[['ID','Sex','Age']])

# Calcular Q1, Q3 e IQR - Masculino
q1 = df_atletas_eventos_m['Age'].quantile(0.25)
mean = df_atletas_eventos_m['Age'].median()
q3 = df_atletas_eventos_m['Age'].quantile(0.75)
iqr = q3 - q1 # Intervalo Interquartil

# Calcular os limites do IQR - Masculino
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

# Identificar os outliers - Masculino
outliers_superior = df_atletas_eventos_m[(df_atletas_eventos_m['Age'] > limite_superior)]
outliers_inferior = df_atletas_eventos_m[(df_atletas_eventos_m['Age'] < limite_inferior)]

# Exibir os resultados - Masculino
print(f"MASCULINO\nQ1: {q1}, \nMedian: {mean}, \nQ3: {q3}, \nIQR: {iqr}")
print(f"Limite Inferior: {limite_inferior}, \nLimite Superior: {limite_superior}")
#print("Outliers:")
print(outliers_inferior[['ID','Sex','Age']])
print(outliers_superior[['ID','Sex','Age']])
```

```python
import pandas as pd

# Carregando dados de um arquivo CSV
df_atletas_eventos = pd.read_csv("https://github.com/ccalmendra/ciencia-dados/releases/download/lcd-v1/m3-atletas-eventos-olimpiadas.csv")

# Exibindo as primeiras linhas
print(df_atletas_eventos.head())

#Exibindo estatísticas básicas dos campos numéricos
df_atletas_eventos.describe()
```

```python
df = df_atletas_eventos
```

```python
freq_relativa_contagem = df["Age"].value_counts()
freq_relativa_percentu = df["Age"].value_counts(normalize=True)
print("Frequência Relativa - contagem absoluta\n", freq_relativa_contagem)
print("Frequência Relativa - percentual\n", freq_relativa_percentu)
```

```python
total_atletas = len(df)
#total unicos seria df['ID'].nunique()
atletas_c_med = df[df["Medal"].notna()] #df dos medalhistas

atletas_c_med = len(atletas_c_med)
print(f"Probabilidade de ganhar uma medalha: {(atletas_c_med / total_atletas):.2%}")

#dropna remove, notna devolve filtrado
```

```python
medalhas_por_edicao = df.groupby("Games")['Medal'].count()
print(medalhas_por_edicao)
```

```python
atletas_por_edicao = df.groupby("Games")['ID'].count()
print(atletas_por_edicao)
```

```python
prob_med_por_ed = medalhas_por_edicao/atletas_por_edicao
print(prob_med_por_ed*100)
```

```python
total_atletas = len(df)
#total unicos seria df['ID'].nunique()
atletas_c_med = df[df["Medal"].notna()] #df dos medalhistas

p_a = len(atletas_c_med)/total_atletas

#dado time brasil

time = 'Brazil'
atletas_time = len(df[df["Team"] == time])

p_b = atletas_time/total_atletas #p atleta ser do br

p_a_e_b = len(atletas_c_med[atletas_c_med['Team']== time])/total_atletas #P(A ∩ B): atleta com medalha e sendo do brasil

# Calcular P(B | A)
p_b_dado_a = p_a_e_b / p_a #prob dele ser brazil dado que ganhou medalha

# Calcular P(A | B)
p_a_dado_b = p_a_e_b / p_b #prob dele ganhar medalha dado que brazil

print(f"Probabilidade de ganhar uma medalha: {p_a}")
print(f"Probabilidade de pertencer ao time: {p_b}")
print(f"Probabilidade de ganhar uma medalha e pertencer ao time: {p_a_e_b}")
print(f"Probabilidade de ganhar medalha dado que pertence ao time {time}: {p_a_dado_b}")
print(f"Probabilidade de ser do time {time} dado que ganhou medalha: {p_b_dado_a}")
```

```python
df_m = df[df["Sex"] == "M"]
p_m = len(df_m)/ total_atletas
print("prob de ser homem: ", p_m)
p_f = 1 - p_m
print("prob de ser mulher: ", p_f)

df_medalhas_f = df[df["Medal"].notna()]
df_medalhas_f =df_medalhas_f[df_medalhas_f["Sex"] == "F"]

qtd_mulheres = total_atletas - len(df_m)
qtd_mulheres_medalhas = len(df_medalhas_f) #quantidade de medalhas totais

prob_med_sendo_mulher = qtd_mulheres_medalhas/qtd_mulheres

print(prob_med_sendo_mulher)
```

```python
# Filtrando participações do Brasil até 2016
df_brasil = df[(df["Team"] == "Brazil") & (df["Year"] <= 2016)]

# Total de participações (registros)
total_participacoes_br = len(df_brasil)

# Total de medalhas ganhas
total_medalhas_br = len(df_brasil[df_brasil["Medal"].notna()])

# Probabilidade P(Medalha | Brasil)
prob_medalha_br = total_medalhas_br / total_participacoes_br

print(f"Probabilidade histórica de medalha (Brasil): {prob_medalha_br:.4%}")
```

```python
# Participações do Brasil apenas em 2016
df_2016 = df[(df["Team"] == "Brazil") & (df["Year"] == 2016)]
n_2016 = len(df_2016)

# Estimativa baseada no histórico
estimativa_2016 = n_2016 * prob_medalha_br

print(f"Número de participações em 2016: {n_2016}")
print(f"Estimativa de medalhas para 2016: {estimativa_2016:.2f}")
```

```python
# Medalhas reais em 2016
real_2016 = len(df_2016[df_2016["Medal"].notna()])

print(f"Resultado Real de medalhas em 2016: {real_2016}")

# Diferença entre Realidade e Estimativa
diferenca = real_2016 - estimativa_2016
print(f"O Brasil ganhou {abs(diferenca):.2f} medalhas a {'mais' if diferenca > 0 else 'menos'} que o estimado.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Carregando dados de um arquivo CSV
df = pd.read_csv("https://github.com/ccalmendra/ciencia-dados/releases/download/lcd-v1/m3-brasileirao-2024.csv")

# Renomear as colunas do DataFrame
df = df.rename(columns={
    df.columns[0]: 'Equipe',
    df.columns[1]: 'Contratacoes',
    df.columns[2]: 'Valores',
    df.columns[3]: 'Pontuacao'
})

# Exibindo as primeiras linhas
print(df.head())

# Criar o gráfico de dispersão
plt.figure(figsize=(10, 6))
plt.scatter(df['Valores'], df['Pontuacao'])

# Adicionar rótulos para cada ponto com o nome da equipe
for i, txt in enumerate(df['Equipe']):
    plt.annotate(txt, (df['Valores'][i], df['Pontuacao'][i]), fontsize=8)

# Adicionar rótulos aos eixos e título ao gráfico
plt.xlabel('Valores')
plt.ylabel('Pontuação')
plt.title('Gráfico de Dispersão: Investimento (R$ Milhões) vs Pontuação')

# Calcular a linha de tendência
z = np.polyfit(df['Valores'], df['Pontuacao'], 1) # MELHOR RETA que passa pelos pontos, com grau 1
p = np.poly1d(z) #vira um polinomio
plt.plot(df['Valores'],p(df['Valores']),"r--") # linha vermelha pontilhada nessa equação

# Exibir o gráfico
plt.grid(True)
plt.show()
```
