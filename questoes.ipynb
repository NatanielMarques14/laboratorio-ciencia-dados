Q2
Escreva uma função soma_impares(lista: list) -> tuple[int, list] que retorne a soma dos números ímpares e a lista com eles. Teste com [3, 8, 7, 2, 11, 6].

```python
def soma_impares(lista:list)->tuple[int,list]:
    soma_impar = 0
    lista_impar = []
    for i in range (0, len(lista)):
        if lista[i] % 2 != 0:
            soma_impar+=lista[i]
            lista_impar.append(lista[i])
    return soma_impar, lista_impar

lista=[3,8,7,2,11,6]
soma_impares(lista)
```

Q12
Com nltk e Counter, carregue o texto da URL de spam da revisão, tokenize com word_tokenize e imprima as 10 palavras mais frequentes. Use try/except para tratar erros de requisição.

```python
import nltk
from nltk.tokenize import word_tokenize
from collections import Counter

nltk.download("punkt")
nltk.download("punkt_tab")

url = "https://raw.githubusercontent.com/ccalmendra/ciencia-dados/refs/heads/main/dados/m2-spam.txt"
response = requests.get(url)
try:
    texto = response.text
    tokens = word_tokenize(texto.lower())
    contagem_palavras=Counter(tokens)
    top10 = contagem_palavras.most_common(10)
    for palavra, freq in top10:
        print(f"{palavra}: {freq}")
except requests.exceptions.RequestException as e:
    print(f"Erro ao carregar o texto da URL: {e}")



```
