A média $$\mu$$ e a variância $$\sigma^2$$ são estatísticas descritivas fundamentais.

- A média é dada por:

\begin{equation}
\mu = \frac{1}{N} \sum_{i=1}^{N} x_i
\end{equation}

- A variância é calculada como:

\begin{equation}
\sigma^2 = \frac{1}{N} \sum_{i=1}^{N} (x_i - \mu)^2
\end{equation}

>**Nota:** A variância mede a dispersão dos dados em relação à média. Um valor alto de variância indica que os dados estão espalhados, enquanto um valor baixo indica que os dados estão mais concentrados.

>**E2.4 - Escreva:** Descreva abaixo usando LaTeX a fórmula do *desvio padrão de uma variável aleatória discreta* (https://pt.wikipedia.org/wiki/Desvio_padr%C3%A3o). Veja os símbolos matemáticos aqui: https://en.wikibooks.org/wiki/LaTeX/Mathematics

\begin{equation}
\sigma = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (x_i - \mu)^2}
\end{equation}

## Prática final - Copa da Computação

10 equipes de futebol participaram da Copa da Computação, um torneio de pontos corridos com apenas um turno. Os placares dos 45 jogos está no arquivo `m1-placares-jogos.csv`.

**Observação:** essa é uma prática de revisão de Python. Evite usar IA generativa, e **busque resolver sozinha(o)**. Veja os exemplos de códigos acima nesse módulo. Busque na documentação oficial ou em sites especializados.

## Objetivos
1. Ler os dados do arquivo `m1-placares-jogos.csv`.

2. Calcular a tabela de classificação da copa seguindo as seguintes regras:

  a. Vitória = 3 pontos p/ vencedor

  b. Empate = 1 ponto p/ ambos

  c. Critério 1: maior pontuador

  d. Critério 2: maior saldo de gols

  e. Critério 3: mais gols feitos

3. Imprimir a tabela de classificação ordenada.



```python
import requests
import csv
import io

# URL do arquivo CSV
url = 'https://raw.githubusercontent.com/ccalmendra/ciencia-dados/refs/heads/main/dados/m1-placares-jogos.csv'

# Fazer a requisição HTTP para obter o conteúdo do arquivo
response = requests.get(url)
csv_data = io.StringIO(response.text)

# Leia o arquivo e carregue em um dicionário para os jogos
jogos = {}

leitor = csv.DictReader(csv_data, delimiter=';')

for i, linha in enumerate(leitor):
    jogos[i] = {
        "Equipe 1": linha["Equipe 1"],
        "Placar 1": int(linha["Placar 1"]),
        "Equipe 2": linha["Equipe 2"],
        "Placar 2": int(linha["Placar 2"])
    }

# Inicialize um dicionário para a tabela
tabela = {}

for jogo in jogos.values():

    # Extrair informações do jogo
    equipe1 = jogo["Equipe 1"]
    equipe2 = jogo["Equipe 2"]
    placar1 = jogo["Placar 1"]
    placar2 = jogo["Placar 2"]

    # Inicializar equipes na tabela
    if equipe1 not in tabela:
        tabela[equipe1] = {
            "Pontos": 0,
            "Gols Feitos": 0,
            "Gols Sofridos": 0,
            "Saldo de Gols": 0
        }

    if equipe2 not in tabela:
        tabela[equipe2] = {
            "Pontos": 0,
            "Gols Feitos": 0,
            "Gols Sofridos": 0,
            "Saldo de Gols": 0
        }

    # Atualizar gols feitos e sofridos
    tabela[equipe1]["Gols Feitos"] += placar1
    tabela[equipe1]["Gols Sofridos"] += placar2

    tabela[equipe2]["Gols Feitos"] += placar2
    tabela[equipe2]["Gols Sofridos"] += placar1

    # Atualizar saldo de gols
    tabela[equipe1]["Saldo de Gols"] = (
        tabela[equipe1]["Gols Feitos"] -
        tabela[equipe1]["Gols Sofridos"]
    )

    tabela[equipe2]["Saldo de Gols"] = (
        tabela[equipe2]["Gols Feitos"] -
        tabela[equipe2]["Gols Sofridos"]
    )

    # Verificar vencedor
    if placar1 > placar2:
        tabela[equipe1]["Pontos"] += 3

    elif placar2 > placar1:
        tabela[equipe2]["Pontos"] += 3

    else:
        tabela[equipe1]["Pontos"] += 1
        tabela[equipe2]["Pontos"] += 1

# Ordenar tabela
classificacao = sorted(
    tabela.items(),
    key=lambda item: (
        item[1]["Pontos"],
        item[1]["Saldo de Gols"],
        item[1]["Gols Feitos"]
    ),
    reverse=True
)

# Imprimir classificação
print("CLASSIFICAÇÃO FINAL\n")

for posicao, (time, dados) in enumerate(classificacao, start=1):

    print(
        f"{posicao}º {time} | "
        f"{dados['Pontos']} pts | "
        f"SG: {dados['Saldo de Gols']} | "
        f"GF: {dados['Gols Feitos']}"
    )
```
